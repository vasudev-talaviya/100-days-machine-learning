```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv("Salary_dataset.csv")
```

```python
data.head()
```

### Remove unuseful column

```python
data.drop(columns='Unnamed: 0',inplace=True,axis=1)
```

```python
data.head()
```

```python
data.info()
```

```python
data.describe()
```

```python
sub , ax = plt.subplots(1,2,figsize=(10,5))

ax[0].set_title("YearExperience and Salary")
ax[0].scatter(data['YearsExperience'],data['Salary'])
ax[0].axhline(data['Salary'].mean(),color='red')
ax[0].set_xlabel("YearsExperience")
ax[0].set_ylabel("Salary")

ax[1].set_title("Salary and YearExperience")
ax[1].scatter(data['Salary'],data['YearsExperience'])
ax[1].axhline(data['YearsExperience'].mean(),color='red')
ax[1].set_xlabel("Salary")
ax[1].set_ylabel("YearsExperience")


plt.tight_layout()
plt.show()
```
